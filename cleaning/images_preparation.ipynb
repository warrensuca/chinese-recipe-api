```python
import pandas as pd
```

```python
the_woks_of_life_path = '../scraping/the_woks_of_life/image_links.csv'
omnivores_cookbook_path = '../scraping/omnivores_cookbook/image_links.csv'

df_wol = pd.read_csv(the_woks_of_life_path)
df_oc = pd.read_csv(omnivores_cookbook_path)
```

```python
df_wol.rename(columns={'Source': 'source'}, inplace=True)
```

```python
df_wol.isna().sum()
```

```python
df_oc.isna().sum()
```

```python
df_wol.fillna('None')
df_oc.fillna('None')

df = pd.concat([df_wol, df_oc], ignore_index=True)
```

```python
df['url'] = df['url'].str.lower()
```

```python
df
```

```python
output_file = "cleaned_image_links.csv"

df.to_csv(
    output_file,
    index=False,
    encoding="utf-8"
)

print(f"Output file: {output_file}")
```
